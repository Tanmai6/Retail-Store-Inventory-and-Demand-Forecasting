```python
import pandas as pd

# Load dataset from data folder
df = pd.read_csv("data/train.csv", index_col=0)

# Cardinality threshold
CARDINALITY_THRESHOLD = 10

# Find categorical columns
categorical_cols = df.select_dtypes(include=["object"]).columns

low_card_cols = []
high_card_cols = []

# Separate columns by cardinality
for col in categorical_cols:
    num_unique = df[col].nunique()

    if num_unique <= CARDINALITY_THRESHOLD:
        low_card_cols.append(col)
    else:
        high_card_cols.append(col)

print("--- CARDINALITY ANALYSIS ---")
print(f"Low Cardinality (One-Hot): {low_card_cols}")
print(f"High Cardinality (Frequency): {high_card_cols}\n")

# Create copy
df_encoded = df.copy()

# Frequency Encoding
for col in high_card_cols:
    freq_map = df_encoded[col].value_counts().to_dict()

    df_encoded[f"{col}_freq"] = df_encoded[col].map(freq_map)

    df_encoded.drop(columns=[col], inplace=True)

# One-Hot Encoding
df_encoded = pd.get_dummies(
    df_encoded,
    columns=low_card_cols,
    dtype=int
)

# Save transformed dataset
output_path = "data/transformed_train.csv"
df_encoded.to_csv(output_path, index=False)

print("--- TRANSFORMATION COMPLETE ---")
print(f"Original shape: {df.shape}")
print(f"Encoded shape: {df_encoded.shape}")
print(f"Saved to: {output_path}")

display(df_encoded.head(3))
```
