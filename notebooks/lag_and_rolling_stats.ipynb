```python
import pandas as pd
import numpy as np
```

```python
df_encoded = pd.read_csv("../data/transformed_train.csv")
group = df_encoded.groupby(["Store ID", "Product ID"])

# =========================================================
# 1. DEMAND
# =========================================================
df_encoded["lag_demand_1"] = group["Demand"].shift(1)
df_encoded["lag_demand_7"] = group["Demand"].shift(7)
df_encoded["lag_demand_14"] = group["Demand"].shift(14)
df_encoded["lag_demand_28"] = group["Demand"].shift(28)

df_encoded["roll_demand_mean_7"] = (
    group["Demand"]
    .shift(1)
    .rolling(7)
    .mean()
    .reset_index(level=[0,1], drop=True)
)

df_encoded["roll_demand_std_7"] = (
    group["Demand"]
    .shift(1)
    .rolling(7)
    .std()
    .reset_index(level=[0,1], drop=True)
)

df_encoded["roll_demand_mean_28"] = (
    group["Demand"]
    .shift(1)
    .rolling(28)
    .mean()
    .reset_index(level=[0,1], drop=True)
)


# =========================================================
# 2. INVENTORY and UNITS SOLD
# =========================================================
df_encoded["lag_inventory_1"] = group["Inventory Level"].shift(1)
df_encoded["roll_inventory_mean_7"] = (
    group["Inventory Level"]
    .shift(1)
    .rolling(7)
    .mean()
    .reset_index(level=[0,1], drop=True)
)

df_encoded["lag_units_sold_1"] = group["Units Sold"].shift(1)
df_encoded["roll_units_sold_mean_7"] = (
    group["Units Sold"]
    .shift(1)
    .rolling(7)
    .mean()
    .reset_index(level=[0,1], drop=True)
)

df_encoded["lag_units_ordered_1"] = group["Units Ordered"].shift(1)
df_encoded["roll_units_ordered_mean_7"] = (
    group["Units Ordered"]
    .shift(1)
    .rolling(7)
    .mean()
    .reset_index(level=[0,1], drop=True)
)


# =========================================================
# 3. PRICE / DISCOUNT / COMPETITION
# =========================================================
df_encoded["lag_price_1"] = group["Price"].shift(1)
df_encoded["roll_price_mean_7"] = (
    group["Price"]
    .shift(1)
    .rolling(7)
    .mean()
    .reset_index(level=[0,1], drop=True)
)

df_encoded["lag_discount_1"] = group["Discount"].shift(1)
df_encoded["roll_discount_mean_7"] = (
    group["Discount"]
    .shift(1)
    .rolling(7)
    .mean()
    .reset_index(level=[0,1], drop=True)
)

df_encoded["lag_competitor_price_1"] = group["Competitor Pricing"].shift(1)


# =========================================================
# 4. EVENT FLAGS
# =========================================================
df_encoded["roll_promotion_7"] = (
    group["Promotion"]
    .shift(1)
    .rolling(7)
    .mean()
    .reset_index(level=[0,1], drop=True)
)

df_encoded["roll_epidemic_7"] = (
    group["Epidemic"]
    .shift(1)
    .rolling(7)
    .mean()
    .reset_index(level=[0,1], drop=True)
)

df_encoded["roll_epidemic_14"] = (
    group["Epidemic"]
    .shift(1)
    .rolling(14)
    .mean()
    .reset_index(level=[0,1], drop=True)
)

df_encoded["roll_epidemic_28"] = (
    group["Epidemic"]
    .shift(1)
    .rolling(28)
    .mean()
    .reset_index(level=[0,1], drop=True)
)
```
